```python
from keras.models import Sequential
from keras.layers import Dense, Flatten
import tensorflow as tf

import matplotlib.pyplot as plt
```

```python
def imgshow(img):
    plt.close()
    plt.imshow(img)
    plt.show()
```

```python
fashion = tf.keras.datasets.fashion_mnist
```

```python
(train_images, train_labels), (test_images, test_labels) = fashion.load_data()
```

```python
train_images.shape
```

```python
print(train_labels[0])
imgshow(train_images[0])
```

```python
class_names = ['T-shirt', 'trouser', 'pullover', 'Dress', 'Cout',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankel boot']
```

```python
train_images = train_images / 255
train_labels = train_labels / 255
```

```python
model = Sequential()
model.add(Flatten(input_shape= (28,28)))
model.add(Dense(128, activation= 'relu'))
model.add(Dense(10))

model.compile(loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits= True), optimizer= 'adam', metrics=['accuracy'])

model.summary()
```

```python
from keras_visualization import visualizer
visualizer(model, file_name= 'DL_model',  file_format= 'png')
```

```python
h = model.fit(train_images, train_labels, epochs= 20)
```

```python
y_pred = model.predict(test_images)
y_pred[0]
```
